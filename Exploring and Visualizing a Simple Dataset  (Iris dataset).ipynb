```python
import pandas as pd                                             # Importing Libraries
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data:pd.DataFrame=pd.read_csv(r"C:\Users\fa440\Downloads\csv files (Data sets)\Iris.csv")      # import dataset using pandas
```

```python
data.shape                                 # to identify the dimensions of data structures like arrays and tables (e.g. (rows, columns)
```

```python
data.columns                                     # Check Columns name
```

```python
data.head()                              # Check data (well-structured or not)
```

```python
                                 # Create basic visualizations:
          # 1- Scatter plot to analyze relationships between variables.
sns.scatterplot(x='SepalLengthCm', y='SepalWidthCm',hue='Species', data=data)

plt.title("Scatter Plot")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.show()
```

```python
                           # 2- Histogram to examine data distribution.
plt.hist(data['PetalLengthCm'])

plt.title("Histogram of Petal Length")
plt.xlabel("Petal Length")
plt.ylabel("Frequency")
plt.show()
```

```python
 # 3- Box plot to detect outliers and spread of values
sns.boxplot(x='Species', y='PetalWidthCm', data=data)

plt.title("Box Plot")
plt.xlabel("Species")
plt.ylabel("Petal Width")
plt.show()
```
